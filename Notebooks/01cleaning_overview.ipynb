## Ho visto che ci sono state delle fusioni tra comuni, questo ne influenza la superficie? In che modo?

```python
# Controllo della variazione della superficie nel tempo
municipalities_changed_name = check_names[check_names > 1].index

surface_changes = (
    df_final_clean[
        df_final_clean["ID_COMUNE"].isin(municipalities_changed_name)].groupby("ID_COMUNE")["SUPERFICIE (KMQ)"]
    .nunique())

# Visualizza solo i comuni per i quali la superficie è cambiata
surface_changes[surface_changes > 1].sort_values(ascending=False).value_counts()
```

La superficie è cambiata per alcuni ID comunali 4, 3 oppure 2 volte

**Quali sono questi comuni**

```python
# ID dei comuni per i quali la superficie cambia nel tempo
mun_changed_sup = surface_changes[surface_changes > 1].index

df_final_clean[
    df_final_clean["ID_COMUNE"].isin(mun_changed_sup)][["ID_COMUNE",
                                                        "COMUNE",
                                                        "TIME_PERIOD",
                                                        "SUPERFICIE (KMQ)",
                                                        "RESIDENTI"]].sort_values(
                                                        ["ID_COMUNE", "TIME_PERIOD"])
```

In alcuni comuni si può notare come la superficie sia cambiata per un piccolo valore

**Di quanto cambia la superficie?**

```python
df_final_clean[df_final_clean["ID_COMUNE"].isin(mun_changed_sup)][["ID_COMUNE","COMUNE",
                                                                "TIME_PERIOD",
                                                                "SUPERFICIE (KMQ)",
                                                                "RESIDENTI"]].sort_values(["ID_COMUNE", "TIME_PERIOD"])
```

```python
surface_variation = (df_final_clean[df_final_clean["ID_COMUNE"].isin(municipalities_changed_name)][
                                            ["ID_COMUNE", "COMUNE", "TIME_PERIOD", "SUPERFICIE (KMQ)"]]
                                                .sort_values(["ID_COMUNE", "TIME_PERIOD"]))

# Superficie minima e massima per ogni comune
surface_range = (surface_variation.groupby("ID_COMUNE").agg(SUPERFICIE_MIN=("SUPERFICIE (KMQ)", "min")
                                                            ,SUPERFICIE_MAX=("SUPERFICIE (KMQ)", "max")))

# Differenza assoluta
surface_range["DIFFERENZA_KMQ"] = (surface_range["SUPERFICIE_MAX"]- surface_range["SUPERFICIE_MIN"])

surface_range.sort_values("DIFFERENZA_KMQ",ascending=False).head(10)
```

Nella maggior parte dei cambi di denominazione la superficie rimane sostanzialmente stabile; solo 2 comuni cambiano classe dimensionale sulla base della superficie minima e massima osservate.
