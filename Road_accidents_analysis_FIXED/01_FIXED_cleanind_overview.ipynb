```python
df_final_raw= pd.merge(df_istat_raw, df_situas_raw, left_on=["REF_AREA","TIME_PERIOD"], right_on=["Codice Comune (numerico)","TIME_PERIOD"], how="left", indicator=True)

df_final_raw
```

```python
left_only=df_final_raw[df_final_raw["_merge"]=="left_only"]

display(left_only)

display(f"Ci sono {left_only.shape[0]} rows che non hanno trovato corrispondenza nel df situas, \n sono comuni che hanno subito variazioni amministrative?")
```

```python
#prendo codice di un comune a caso e vedo qual è il primo anno di NON corrispondenza
codice = left_only["REF_AREA"].iloc[0]
anno = left_only["TIME_PERIOD"].iloc[0]

print(codice, anno)
```

```python
#guardo se esisteva in df situas --->non esisteva
df_situas_raw[(df_situas_raw["Codice Comune (numerico)"] == codice) &
                (df_situas_raw["TIME_PERIOD"] == anno)]
```

```python
#guardo quel codice preso come esempio fino a che anno è stato registrato
df_situas_raw[df_situas_raw["Codice Comune (numerico)"] == codice][["Codice Comune (numerico)", "Comune", "TIME_PERIOD"]].sort_values("TIME_PERIOD")
```

```python
#creo due set di codici univoci dei due df in maniera tale da sottrarli e vedere se istat ha + o - codici
# rispetto a situas
codici_istat = set(df_istat_raw["REF_AREA"].unique())
codici_situas = set(df_situas_raw["Codice Comune (numerico)"].unique())

print(len(codici_istat - codici_situas))
```

```python
df_situas_raw[
    df_situas_raw["Comune"].str.contains(
        "LU E CUCCARO",
        case=False,
        na=False
    )
][["Comune",
   "Codice Comune (numerico)",
   "TIME_PERIOD"]].sort_values("TIME_PERIOD")
```

```python

primo_left = (
    left_only
    .groupby("REF_AREA")["TIME_PERIOD"]
    .min()
    .reset_index()
    .rename(columns={
        "REF_AREA": "Codice Comune (numerico)",
        "TIME_PERIOD": "Primo left_only"
    })
)

ultimi_anni = (
    df_situas_raw[
        df_situas_raw["Codice Comune (numerico)"].isin(left_only["REF_AREA"])
    ]
    .groupby("Codice Comune (numerico)")["TIME_PERIOD"]
    .max()
    .reset_index()
)

controllo = ultimi_anni.merge(
    primo_left,
    on="Codice Comune (numerico)"
)

controllo["Differenza"] = (
    controllo["Primo left_only"] -
    controllo["TIME_PERIOD"]
)

controllo.sort_values("Differenza")
controllo["Differenza"].value_counts()
```

La merge è stata effettuata sulla chiave composta (REF_AREA, TIME_PERIOD) e validata tramite validate="many_to_one". Le osservazioni senza corrispondenza (left_only) sono risultate riconducibili quasi esclusivamente a variazioni amministrative dei comuni. Per 283 dei 284 codici comunali coinvolti, il primo anno senza corrispondenza coincide con l'anno successivo all'ultima presenza nel dataset SITUAS, indicando fusioni, soppressioni o cambi di codice. È stato individuato un solo caso anomalo, oggetto di verifica puntuale.

```python
#Creo il df con solo le colonne di interesse

df_final_clean= df_final_raw[["REF_AREA","DATA_TYPE","RESULT","TIME_PERIOD","OBS_VALUE","Codice Regione", "Codice Provincia/Uts",
                      "Codice Comune (numerico)",
                      "Comune","Capoluogo di Provincia/Uts", "Capoluogo di Regione", "Superficie (Kmq)", 
                      "Popolazione residente", "Anno (Popolazione residente)", "_merge"]]
df_final_clean
```

```python
#voglio vedere quanti comuni univoci c' erano nel 2020
df_final_clean[df_final_clean["TIME_PERIOD"] == 2020]["REF_AREA"].nunique()
```

```python
#PRIMA DI ELIMINARE LA COLONNA MERGE MI RICREO IL DF CON SOLO QUELLI CHE SI SONO MATCHATI
#SI VANNO A PERDERE SOLO 850 RECORD CIRCA SU PIù DI 500000

df_final_clean=df_final_clean[df_final_clean["_merge"] == "both"].copy()
```

```python
#Elimino merge che non mi serve più, il codice d
df_final_clean=df_final_clean.drop(columns=["_merge","Anno (Popolazione residente)","Codice Comune (numerico)"]) #ho dovuto riscrivere df_final_clean= perchè drop non eliminerebbe 
                                                            #in maniera permanente la colonna, se non volessi salvare il 
                                                            #df come un nuovo df, dovrei mettere drop(columns=["_merge"], inplace=True)

df_final_clean
```

```python
df_final_clean.isna().sum()
```

```python
#controllo i tipi di dato
df_final_clean.info()
```

```python
#SISTEMO I TIPI DELLE COLONNE

df_final_clean["Codice Regione"] = (df_final_clean["Codice Regione"].astype("Int64"))
# Int64 (con la I maiuscola) permette di avere interi e valori mancanti nella stessa colonna.
df_final_clean["Codice Provincia/Uts"] = (df_final_clean["Codice Provincia/Uts"].astype("Int64"))
df_final_clean["Capoluogo di Provincia/Uts"] = (df_final_clean["Capoluogo di Provincia/Uts"].astype("Int64"))
df_final_clean["Capoluogo di Regione"] = (df_final_clean["Capoluogo di Regione"].astype("Int64"))
df_final_clean["Popolazione residente"] = (df_final_clean["Popolazione residente"].astype("Int64"))
```

```python
df_final_clean[df_final_clean["Superficie (Kmq)"].str.contains("\.", regex=True)] #così mi dà come risultato tutto il df con la colonna sup con i caratteri contenenti .

df_final_clean[df_final_clean["Superficie (Kmq)"].str.contains("\.", regex=True)]["Superficie (Kmq)"] #così mi fa vedere solo la colonna Superficie con i caratteri contenti .
#N.B. ricorda lo \ prima del . altrimenti cercherebbe una concatenazione di parole e non proprio il punto come valore

#regex=False -->Interpreta la stringa come testo letterale, quindi i caratteri speciali sono esattamente quello che sono 
                #a livello letterale. Serve per avere una stringa esatta
#regex=True --> quando vuoi sfruttare i simboli delle espressioni regolari (\s (spazio), \d, ^, $, ., +, *, ecc.) e 
                # non che sia letterale il carattere speciale
```

```python
df_final_clean["Superficie (Kmq)"] = (df_final_clean["Superficie (Kmq)"].str.replace(".", "", regex=False)
                                        .str.replace(",", ".", regex=False)
                                        .astype(float))
```

```python
#Rinomino i valori per renderli più chiari
df_final_clean["DATA_TYPE"] = df_final_clean["DATA_TYPE"].replace({"KILLINJ": "morti_e_feriti",
                                                                    "ROADACC": "incidenti_stradali"})

df_final_clean["RESULT"] = df_final_clean["RESULT"].replace({"M": "morti",
                                                            "F": "feriti",
                                                            "9": "incidenti"})

df_final_clean=df_final_clean.rename(columns={"REF_AREA" : "ID_COMUNE", 
                                                              "Codice Provincia/Uts" : "ID_PROVINCIA",
                                                              "Codice Regione" : "ID_REGIONE",
                                                              "Capoluogo di Provincia/Uts": "CAPOLUOGO PROVINCIA",
                                                              "Capoluogo di Regione" : "CAPOLUOGO REGIONE",
                                                              "Comune" : "COMUNE",
                                                              "Popolazione residente" : "RESIDENTI",
                                                              "Superficie (Kmq)" : "SUPERFICIE (KMQ)",})
df_final_clean
```

```python
#AGGIUNGO IL NOME DELLE REGIONI

regioni = {
    1: "Piemonte",
    2: "Valle d'Aosta",
    3: "Lombardia",
    4: "Trentino-Alto Adige",
    5: "Veneto",
    6: "Friuli-Venezia Giulia",
    7: "Liguria",
    8: "Emilia-Romagna",
    9: "Toscana",
    10: "Umbria",
    11: "Marche",
    12: "Lazio",
    13: "Abruzzo",
    14: "Molise",
    15: "Campania",
    16: "Puglia",
    17: "Basilicata",
    18: "Calabria",
    19: "Sicilia",
    20: "Sardegna"}

df_final_clean["REGIONE"] = df_final_clean["ID_REGIONE"].map(regioni)
```
